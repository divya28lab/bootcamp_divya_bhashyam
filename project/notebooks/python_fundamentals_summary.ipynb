# Python Fundamentals Summary

This notebook demonstrates basic Python, NumPy, and pandas operations using mock market data for the Market Early-Warning System.

```python
# Basic Python operations

market_name = "Market Early-Warning System"
days = 5
volatility = [0.12, 0.15, 0.18, 0.14, 0.21]

print("Project:", market_name)
print("Number of observations:", days)
print("Average volatility:", sum(volatility) / len(volatility))
```

```python
import numpy as np

# Mock market indicators
volatility = np.array([0.12, 0.15, 0.18, 0.14, 0.21])
returns = np.array([0.01, -0.02, 0.015, -0.03, 0.005])

print("Volatility:", volatility)
print("Average volatility:", np.mean(volatility))
print("Maximum volatility:", np.max(volatility))
print("Average return:", np.mean(returns))
```

```python
import pandas as pd

data = {
    "day": [1, 2, 3, 4, 5],
    "volatility": [0.12, 0.15, 0.18, 0.14, 0.21],
    "return": [0.01, -0.02, 0.015, -0.03, 0.005]
}

df = pd.DataFrame(data)

print(df)

print("\nAverage volatility:", df["volatility"].mean())
print("Minimum return:", df["return"].min())
print("Maximum return:", df["return"].max())
```
