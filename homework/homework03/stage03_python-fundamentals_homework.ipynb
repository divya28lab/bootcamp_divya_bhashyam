## 1. NumPy Operations

```python
arr = np.array([1, 2, 3, 4, 5])

squared = arr ** 2
doubled = arr * 2

print("Original array:", arr)
print("Squared:", squared)
print("Doubled:", doubled)
```

## 2. Compare Loop vs Vectorization

```python
arr = np.arange(100000)

def loop_square(arr):
    result = []
    for x in arr:
        result.append(x ** 2)
    return result

def vectorized_square(arr):
    return arr ** 2

%timeit loop_square(arr)
%timeit vectorized_square(arr)
```

## 4. Summary Statistics and Groupby

```python
# Calculate average closing price and total volume by ticker
summary = df.groupby("ticker").agg(
    average_close=("close", "mean"),
    total_volume=("volume", "sum")
).reset_index()

print(summary)

# Save the grouped results
output_path = "/Users/db/Desktop/bootcamp_divya_bhashyam/homework/homework03/data/processed/market_summary.csv"

summary.to_csv(output_path, index=False)

print("Saved to:", output_path)
```
